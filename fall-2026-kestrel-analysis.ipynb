## 1. Introduction and context

### 1.1 Problem: heat exposure and pedestrian behavior

Urban streets can expose pedestrians to intense solar radiation, especially where trees and buildings offer little shade. Two nearby locations may share similar ambient air temperature yet differ in the radiant heat experienced by a person at sidewalk level. Shade is therefore relevant to pedestrian thermal exposure, not merely to the surrounding air temperature.

Observational evidence suggests pedestrians respond to these differences. In downtown Austin, **Kim and Brown (2022)** found that walking, standing, and sitting behavior was associated with micro-scale thermal conditions, including shade, sky-view factor, and tree canopy ([*Science of the Total Environment*](https://doi.org/10.1016/j.scitotenv.2021.152143)). A later review reports that **85% of 256 observed pedestrians** were in shade in that study, although this proportion depends on the particular locations and available shade ([*Buildings and Cities*, 2025](https://doi.org/10.5334/bc.574)). The same 2025 study observed **5,263 pedestrians** at 34 locations in Tel Aviv–Yafo and found approximately **60% in shade**, with usage related to shade availability. These findings indicate behavioral preference under the studied conditions, not a universal pedestrian proportion.

### 1.2 Connection to traffic safety

A nationwide U.S. study identified an association between heat waves and fatal traffic crashes, with stronger positive associations on days involving medium or high solar radiation ([**Wu et al.**, *Accident Analysis & Prevention*, 2018](https://www.sciencedirect.com/science/article/pii/S0001457518303257)). A more recent case-crossover analysis covering **177 California cities (2012–2023)** found elevated severe-crash risk at extreme temperatures, with particularly important findings for intersections and vulnerable road users, including pedestrians ([**Hsu et al.**, 2026](https://doi.org/10.1007/s44327-026-00248-6)). These are observational associations; they do not establish that heat or shade caused an individual crash.

### 1.3 Measurement limitation and proposed gap

Existing traffic-safety analyses have commonly represented heat using ambient temperature at a **regional or city scale**, including **gridded weather products**. For example, Wu et al. used NLDAS-2 gridded air-temperature estimates; Hsu et al. used PRISM daily temperature data at approximately 800 m resolution, population-weighted to cities. **It would be inaccurate to say these studies exclusively used meteorological weather-station readings.** Their exposure measures nevertheless do **not directly resolve whether a specific pedestrian location was shaded at the hour of a crash**. See the [Wu study](https://www.sciencedirect.com/science/article/pii/S0001457518303257) and [Hsu methods](https://pmc.ncbi.nlm.nih.gov/articles/PMC13102899/).

Urban greenery, canopy, sky-view factor, and shade have been examined in pedestrian comfort and urban-climate research ([Kim and Brown, 2022](https://doi.org/10.1016/j.scitotenv.2021.152143); [global pedestrian shade mapping, *Nature Communications*, 2026](https://www.nature.com/articles/s41467-026-69190-w)). The **candidate gap** is their integration as **time- and location-specific pedestrian exposure variables in crash-risk models**. This is a testable proposed contribution, **not a verified claim that no existing study has done so**; a systematic literature review is required before asserting novelty.

### 1.4 Role of the Kestrel pilot

The available Kestrel 5400 field sessions provide a **microclimate pilot**: they can characterize air temperature, globe temperature, wet-bulb globe temperature (WBGT), wind, humidity, and other instrument-reported values under field-note classifications of Sun, Shade, Mixed, or Unknown. This pilot can evaluate measurement behavior and the practical relevance of local exposure differences. **It cannot, by itself, establish any heat–crash relationship.** Crash-risk questions require separate spatially and temporally linked crash, pedestrian-exposure, weather, and shade data.

The [Kestrel 5400 manufacturer documentation](https://kestrelinstruments.com/kestrel-5400-heat-stress-tracker) also distinguishes instrument-sensed variables from internally calculated indicators; a value appearing in the CSV does not necessarily mean it was measured by an independent physical sensor.

## 2. Research objective

**Primary objective:** Evaluate whether incorporating solar radiation and pedestrian-relevant, time-specific shade/urban geometry changes the estimated and/or classified relationship between heat exposure and pedestrian-involved crash risk across Austin's urban roadway network, compared with a model using ambient air temperature alone.

**Supporting pilot objective:** Describe and quality-check the Kestrel 5400 microclimate measurements under field-note Sun, Shade, Mixed, and Unknown conditions, and determine whether they provide useful local evidence of thermal-exposure differences beyond ambient air temperature.

### Research questions

**RQ1 — Ambient baseline:** How is ambient air temperature associated with pedestrian-involved crash occurrence or severity after accounting for available roadway, temporal, and pedestrian-exposure variables?

**RQ2 — Solar radiation:** Does adding solar radiation to the baseline model change estimated heat-related crash associations or improve out-of-sample predictive performance?

**RQ3 — Pedestrian-scale shade:** Does adding time-specific pedestrian-accessible shade or related urban geometry change risk estimates, site classifications, or predictive performance beyond ambient temperature and solar radiation?

**RQ4 — Field pilot:** How do the original Kestrel environmental readings vary across the field-note exposure classifications, and what measurement-quality limitations should be considered before interpreting those differences?

### Working hypotheses (to be tested, not treated as findings)

- **H1:** Higher ambient temperature is associated with increased pedestrian-involved crash risk, conditional on the selected controls. The direction and size remain empirical questions in Austin.
- **H2:** Solar radiation adds explanatory or predictive information beyond ambient air temperature alone.
- **H3:** Time- and location-specific shade/urban-geometry measures add explanatory or predictive information beyond ambient temperature and regional solar radiation; some road-segment risk rankings change.
- **H4 (pilot):** Sessions documented as directly sun-exposed tend to show higher globe temperature and WBGT than shade sessions under otherwise comparable conditions. Because the pilot is observational, these comparisons cannot isolate a causal shade effect without accounting for time, weather, wind, device behavior, and placement.

**Model sequence (future work):** M0 = ambient temperature baseline; M1 = M0 + solar radiation; M2 = M1 + pedestrian-scale shade/geometry. The precise outcome, exposure denominator, spatial unit, time step, covariates, estimation approach, and comparison criteria must be specified before fitting models. If reliable pedestrian counts are unavailable, clearly label results as *crash occurrence or reported crash frequency* rather than exposure-normalized individual pedestrian risk.

### Code cell 5 — Define the file-listing function

**Goal:** Specify how to list genuine CSV files within the original ZIP. **Expected output:** None; defining a function does not read the files.

```python
def list_csv_files_global_function(source_files_local_zip):
    csv_file_names_local_list = [
        file_name_local_text
        for file_name_local_text in source_files_local_zip.namelist()
        if file_name_local_text.lower().endswith(csv_file_extension_global_parameter)
        and not file_name_local_text.startswith(excluded_folder_prefix_global_parameter)
        and excluded_filename_fragment_global_parameter not in file_name_local_text
    ]
    return sorted(csv_file_names_local_list)
```

### Code cell 6 — Define the instrument-ID function

**Goal:** Read the instrument ID from the first three metadata rows of one original CSV file. **Expected output:** None.

```python
def read_instrument_ID_global_function(source_files_local_zip, csv_file_local_name):
    with source_files_local_zip.open(csv_file_local_name) as file_contents_local_binary:
        text_file_local_stream = io.TextIOWrapper(
            file_contents_local_binary,
            encoding=source_text_encoding_global_parameter,
            errors=source_text_errors_global_parameter
        )
        metadata_rows_local_list = list(
            itertools.islice(csv.reader(text_file_local_stream), metadata_rows_global_parameter)
        )

    for metadata_row_local_values in metadata_rows_local_list:
        if (len(metadata_row_local_values) >= 2
                and metadata_row_local_values[0].strip() == instrument_ID_label_global_parameter):
            return metadata_row_local_values[1].strip()
    return None
```

### Code cell 14 — Define function for reading one CSV file

**Goal:** Preserve the original measurements, session name, timestamps, and field notes.

```python
def read_csv_measurements_global_function(source_files_local_zip, csv_file_local_name):
    original_bytes_local_data = source_files_local_zip.read(csv_file_local_name)
    decoded_text_local_value = None
    for encoding_local_name in ["utf-8-sig", "cp1252", "latin-1"]:
        try:
            decoded_text_local_value = original_bytes_local_data.decode(encoding_local_name)
            break
        except UnicodeDecodeError:
            continue
    if decoded_text_local_value is None:
        raise ValueError(f"Cannot decode {csv_file_local_name}")
    file_rows_local_values = list(csv.reader(io.StringIO(decoded_text_local_value, newline="")))
    metadata_local_values = {
        row_local_values[0].strip(): row_local_values[1].strip()
        for row_local_values in file_rows_local_values[:3]
        if len(row_local_values) >= 2
    }
    instrument_ID_local_value = metadata_local_values.get("Serial Number")
    header_position_local_index = next((
        index_local_value for index_local_value, row_local_values
        in enumerate(file_rows_local_values[:15])
        if "FORMATTED DATE_TIME" in row_local_values and "Record name" in row_local_values
    ), None)
    if header_position_local_index is None:
        raise ValueError(f"Missing measurement header: {csv_file_local_name}")
    source_fields_local_names = file_rows_local_values[header_position_local_index]
    units_local_values = dict(zip(source_fields_local_names, file_rows_local_values[header_position_local_index + 1]))
    file_information_local_dictionary = {
        "Original CSV file": csv_file_local_name,
        "Instrument ID": instrument_ID_local_value,
        "SHA-256": hashlib.sha256(original_bytes_local_data).hexdigest(),
        "Air temperature unit": units_local_values.get("Temperature"),
        "Globe temperature unit": units_local_values.get("Globe Temperature"),
        "Wind speed unit": units_local_values.get("Wind Speed")
    }
    measurements_local_rows = []
    for original_row_local_number, original_row_local_values in enumerate(
        file_rows_local_values[header_position_local_index + 2:],
        start=header_position_local_index + 3
    ):
        if not any(value_local_text.strip() for value_local_text in original_row_local_values):
            continue
        if len(original_row_local_values) != len(source_fields_local_names):
            raise ValueError(f"Wrong number of columns in {csv_file_local_name}, row {original_row_local_number}")
        reading_local_fields = dict(zip(source_fields_local_names, original_row_local_values))
        if reading_local_fields.get("Data Type", "").strip().lower() != "session":
            continue
        reading_local_time = pd.to_datetime(
            reading_local_fields.get("FORMATTED DATE_TIME", ""),
            format="%Y-%m-%d %I:%M:%S %p", errors="coerce"
        )
        if pd.isna(reading_local_time):
            raise ValueError(f"Invalid timestamp in {csv_file_local_name}, row {original_row_local_number}")
        reading_local_record = {
            "Original CSV file": csv_file_local_name,
            "Source row": original_row_local_number,
            "Instrument ID": instrument_ID_local_value,
            "Session ID": reading_local_fields.get("Record name", "").strip(),
            "Reading time": reading_local_time,
            "Field Notes": reading_local_fields.get("Notes", ""),
            "Location link": reading_local_fields.get("Location coordinates", ""),
            "Location description": reading_local_fields.get("Location description", ""),
        }
        for source_local_name, display_local_name in original_measurement_names_global_parameter.items():
            reading_local_record[display_local_name] = pd.to_numeric(
                reading_local_fields.get(source_local_name, ""), errors="coerce"
            )
        measurements_local_rows.append(reading_local_record)
    return file_information_local_dictionary, measurements_local_rows
```

### Code cell 16 — Define function for classifying original notes

**Goal:** Use conservative field-note rules; never infer exposure from a high globe temperature.

**Known limitation of the published classification:** "No shade" matches both the sun and shade patterns, so that phrase alone returns Mixed. Preserve this legacy rule to reproduce the published counts. Review the affected notes and regenerate every dependent output before interpreting or publishing revised exposure groups. Mixed can also describe changing cloud cover, not only a change in physical shade.

```python
def classify_field_notes_global_function(original_notes_local_text):
    notes_local_text = re.sub(
        r"^enter notes[\s\-:]*", "", str(original_notes_local_text or "").strip().lower()
    )
    if not notes_local_text:
        return "Unknown"
    mixed_phrases_local_parameter = (
        "partially in shade", "mostly in sun", "most of measurement in sun",
        "sun periodically", "varying cloud", "changes in cloud", "first minute was",
        "started cloudy", "partially cloudy", "cloudy location",
        "cloudiness throughout", "shade nearby"
    )
    if any(phrase_local_text in notes_local_text for phrase_local_text in mixed_phrases_local_parameter):
        return "Mixed"
    sun_local_detected = bool(re.search(
        r"\b(direct sun|direct sunlight|in the sun|sunny|sunniest|unshaded|no shade)\b",
        notes_local_text
    ))
    shade_local_detected = bool(re.search(
        r"\b(shade|shaded|shady|shadow|underneath trees)\b", notes_local_text
    ))
    if sun_local_detected and shade_local_detected:
        return "Mixed"
    if shade_local_detected:
        return "Shade"
    if sun_local_detected:
        return "Sun"
    return "Unknown"
```

### Code cell 23 — Review measurement quality without removing sessions

**Goal:** Flag incomplete, irregular, short, or drifting sessions; keep all 47.

```python
eligible_measurements_local_table = eligible_measurements_local_table.sort_values(
    ["Instrument ID", "Session ID", "Reading time"]
).copy()
eligible_measurements_local_table["Seconds between readings"] = (
    eligible_measurements_local_table.groupby(["Instrument ID", "Session ID"])["Reading time"]
    .diff().dt.total_seconds()
)
quality_review_local_rows = []
for (instrument_ID_local_value, session_ID_local_value), readings_local_group in eligible_measurements_local_table.groupby(
    ["Instrument ID", "Session ID"]
):
    elapsed_local_minutes = readings_local_group["Elapsed minutes"]
    globe_local_values = readings_local_group["Globe temperature (°F)"]
    early_local_mean = globe_local_values.loc[elapsed_local_minutes <= 2].mean()
    late_local_mean = globe_local_values.loc[elapsed_local_minutes >= elapsed_local_minutes.max() - 2].mean()
    intervals_local_seconds = readings_local_group["Seconds between readings"].dropna()
    quality_review_local_rows.append({
        "Instrument ID": instrument_ID_local_value,
        "Session ID": session_ID_local_value,
        "Irregular intervals": int((~intervals_local_seconds.between(4, 6)).sum()),
        "Missing core readings": int(readings_local_group[[
            "Air temperature (°F)", "Globe temperature (°F)",
            "Wet bulb globe temperature (°F)", "Relative humidity (%)", "Wind speed (mph)"
        ]].isna().any(axis=1).sum()),
        "Globe temperature change over session (°F)": late_local_mean - early_local_mean
    })
measurement_quality_review_local_table = eligible_sessions_local_table.merge(
    pd.DataFrame(quality_review_local_rows), on=["Instrument ID", "Session ID"], validate="one_to_one"
)
measurement_quality_review_local_table["Short session"] = measurement_quality_review_local_table["Duration (minutes)"].lt(8)
measurement_quality_review_local_table["Large globe temperature change"] = (
    measurement_quality_review_local_table["Globe temperature change over session (°F)"].abs().gt(5)
)
measurement_quality_review_local_table["Requires review"] = (
    measurement_quality_review_local_table["Short session"]
    | measurement_quality_review_local_table["Irregular intervals"].gt(0)
    | measurement_quality_review_local_table["Missing core readings"].gt(0)
    | measurement_quality_review_local_table["Large globe temperature change"]
)
print("Sessions flagged for review:", int(measurement_quality_review_local_table["Requires review"].sum()))
display(measurement_quality_review_local_table.loc[
    measurement_quality_review_local_table["Requires review"],
    ["Session ID", "Instrument ID", "Original Classification", "Field Notes", "Irregular intervals", "Globe temperature change over session (°F)"]
].round(2))
```

### Code cell 28 — Define function for reading original location links

**Goal:** Extract latitude and longitude without interpolating a heat surface.

```python
def parse_location_link_global_function(original_link_local_text):
    location_match_local_result = re.search(
        r"query=([-+]?\d+(?:\.\d+)?),([-+]?\d+(?:\.\d+)?)",
        str(original_link_local_text)
    )
    if location_match_local_result is None:
        return (np.nan, np.nan)
    return (float(location_match_local_result.group(1)), float(location_match_local_result.group(2)))
```
